```python
df = pd.read_excel("Telco_customer_churn.xlsx")

df['Total Charges'] = pd.to_numeric(
    df['Total Charges'],
    errors='coerce'
)

df['Total Charges'] = df['Total Charges'].fillna(0)

df.head()
```

```python
drop_cols = [
    'CustomerID',
    'Country',
    'State',
    'City',
    'Zip Code',
    'Lat Long',
    'Latitude',
    'Longitude',
    'Count',
    'Churn Reason'
]

df = df.drop(columns=drop_cols)
```

```python
df.columns
```

```python
le = LabelEncoder()

for col in df.select_dtypes(include='object').columns:
    df[col] = le.fit_transform(
        df[col].astype(str)
    )
```

```python
df.head()
```

```python
y = df['Churn Value']
```

```python
X = df.drop(
    columns=[
        'Churn Value',
        'Churn Label'
    ]
)
```

```python
print("Features:", X.shape)
print("Target:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
X.to_csv(
    "customer_features.csv",
    index=False
)

y.to_csv(
    "customer_target.csv",
    index=False
)
```
